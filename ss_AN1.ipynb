```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x  = np.linspace(5,-5,100)
plt.plot(x,1 / (1+np.exp(-x)), label='sigmoid')

plt.xlabel('INPUT')
plt.ylabel('ACTIVATION')
plt.title('Sigmoid Activation Function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.tanh(x), label='tanh')

plt.xlabel('INPUT')
plt.ylabel('ACTIVATION')
plt.title('Hyperbolic Tangent Activation Function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.maximum(0,x), label='ReLu')

plt.xlabel('Input')
plt.ylabel('Activation')
plt.title('Rectified Linear Function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, x, label='identify')
plt.xlabel('INPUT')
plt.ylabel('ACTIVATION')
plt.title('Identify Activation Function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(x, np.exp(x)/np.sum(np.exp(x)), label='Softmax')
plt.xlabel('INPUT')
plt.ylabel('ACTIVATION')
plt.title('Softmax Activation Function')
plt.legend()
plt.grid(True)
plt.show()
```
