```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
X = np.array([[0,0], [0,1], [1,0], [1,1]])
Y = np.array([-1, -1, -1, 1])

w = np.zeros(X.shape[1])
b = 0

# Go through each training point multiple times (6 times total).
#If the perceptron makes a wrong prediction, it updates weights and bias to learn from the mistake.
"""
np.sign(...) returns:
+1 if result > 0
-1 if result < 0
0 if result == 0 (though 0 is rare in practice)
"""

for _ in range(6):
    for i in range(X.shape[0]):
        y_pred = np.sign(np.dot(X[i], w) + b)
        if y_pred != Y[i]:
            w += 0.3 * Y[i] * X[i]
            b += 0.3 * Y[i]


x_min, x_max = X[:, 0].min() -1, X[:, 0].max() + 1 
y_min, y_max = X[:, 1].min() - 1, X[:,1].max() + 1     
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),   # np.arange(start, stop, step)
                     np.arange(y_min, y_max, 0.01))   #-->np.meshgrid(...) turns the 1D arrays into 2D grid coordinates


Z = np.sign(np.dot(np.c_[xx.ravel(), yy.ravel()], w) + b)
Z = Z.reshape(xx.shape)


plt.contourf(xx, yy, Z, alpha = 0.8)
plt.scatter(X[:, 0], X[:, 1], c=Y)
plt.xlabel('X1')
plt.ylabel('X2')
plt.title('Perceptron Decision Regions')
plt.show()
```
